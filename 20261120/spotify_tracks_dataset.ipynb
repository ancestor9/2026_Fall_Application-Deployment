```python
import pandas as pd

# ZIP 파일로부터 CSV 직접 로드
df = pd.read_csv('/content/dataset.csv.zip')

# 데이터프레임 구조 정보 출력
print(df.info())

# 처음 5개 행 표시
display(df.head())
```

```python
def summarize_columns(dataframe):
    summary_data = []
    for col in dataframe.columns:
        null_count = dataframe[col].isnull().sum()
        unique_count = dataframe[col].nunique()
        top_7_values = dataframe[col].value_counts().head(7).to_dict()

        summary_data.append({
            'Column': col,
            'Missing Values': null_count,
            'Unique Values': unique_count,
            'Top 7 Frequent Values': top_7_values
        })

    summary_df = pd.DataFrame(summary_data)
    return summary_df

# 함수 실행 및 결과 출력
result = summarize_columns(df)
pd.set_option('display.max_colwidth', None)
display(result)
```
