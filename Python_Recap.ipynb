### **seaborn의 tips를 불러와 EDA <--- colab gemini**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 한글 폰트 설정 (필요 시 주석 해제하여 사용)
# plt.rc('font', family='NanumBarunGothic')

# 1. tips 데이터셋 로드
tips = sns.load_dataset('tips')

# 2. 데이터 기본 구조 확인
print("=== 데이터셋 기본 정보 ===")
print(tips.info())
print("\n=== 처음 5개 행 데이터 ===")
print(tips.head())

# 3. 수치형 데이터의 기초 통계량 확인
print("\n=== 수치형 데이터 기초 통계량 ===")
print(tips.describe())

# 4. 범주형 데이터의 빈도 확인
print("\n=== 범주형 데이터 빈도 (성별) ===")
print(tips['sex'].value_counts())

# 5. 간단한 시각화 (EDA)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 요일(day)별 팁(tip) 분포
sns.boxplot(x='day', y='tip', data=tips, ax=axes[0], palette='Set2')
axes[0].set_title('Tips Distribution by Day')

# 총 결제 금액(total_bill)과 팁(tip)의 상관 관계
sns.scatterplot(x='total_bill', y='tip', hue='smoker', data=tips, ax=axes[1])
axes[1].set_title('Total Bill vs Tip')

plt.tight_layout()
plt.show()
```

```python
# 처음 10개 샘플을 1차원(1, 784) 한 줄 형태로 시각화
first_10_samples = m_test.head(10)

fig, axes = plt.subplots(10, 1, figsize=(15, 6))

for idx, (i, row) in enumerate(first_10_samples.iterrows()):
    label = row.iloc[0]
    # 28x28로 변형하지 않고 1x784 형태 그대로 유지
    pixels_1d = row.iloc[1:].values.reshape(1, 784)

    axes[idx].imshow(pixels_1d, cmap='gray', aspect='auto')
    axes[idx].set_title(f"Label: {label} (1D Pixel Strip)", fontsize=9, loc='left')
    axes[idx].axis('off')

plt.tight_layout()
plt.show()
```

```python
# 처음 10개 샘플을 2행 5열(정사각형 숫자 형태)로 시각화
first_10_samples = m_test.head(10)

fig, axes = plt.subplots(2, 5, figsize=(10, 5))
axes = axes.flatten()

for idx, (i, row) in enumerate(first_10_samples.iterrows()):
    label = row.iloc[0]
    # 1차원 데이터를 2D 이미지 형태(28, 28)로 복원
    pixels = row.iloc[1:].values.reshape(28, 28)

    axes[idx].imshow(pixels, cmap='gray')
    axes[idx].set_title(f"Label: {label}")
    axes[idx].axis('off')

plt.tight_layout()
plt.show()
```

```python
from skimage import data
import matplotlib.pyplot as plt

# 1. 실제 RGB 예제 이미지 로드 (scikit-image의 고양이 이미지)
cat_image = data.chelsea()

# 2. 데이터 구조 정보 출력
print("=== 실제 RGB 예제 데이터 정보 ===")
print("이미지 차원 (Height, Width, Channels) :", cat_image.shape)
print("데이터 타입 :", cat_image.dtype)
print("최대 픽셀 값 :", cat_image.max(), "/ 최소 픽셀 값 :", cat_image.min())

# 3. 이미지 및 RGB 채널 분리 시각화
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# 원본 이미지
axes[0].imshow(cat_image)
axes[0].set_title("Original RGB Cat")
axes[0].axis("off")

# R 채널
axes[1].imshow(cat_image[:, :, 0], cmap="Reds")
axes[1].set_title("Red Channel")
axes[1].axis("off")

# G 채널
axes[2].imshow(cat_image[:, :, 1], cmap="Greens")
axes[2].set_title("Green Channel")
axes[2].axis("off")

# B 채널
axes[3].imshow(cat_image[:, :, 2], cmap="Blues")
axes[3].set_title("Blue Channel")
axes[3].axis("off")

plt.tight_layout()
plt.show()
```

```python
# @title 단어 Tokenizer
from transformers import AutoTokenizer, AutoModel
import torch

model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)

text = "have the bards who preceded me left any theme unsung"
inputs = tokenizer(text, return_tensors="pt")

with torch.no_grad():
    outputs = model(**inputs)

# outputs.last_hidden_state 구조: (batch_size, sequence_length, hidden_size)
# 각 토큰(단어)별 벡터 표현
word_vectors = outputs.last_hidden_state[0]

# 토큰과 벡터 매핑 예시
tokens = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
for token, vec in zip(tokens, word_vectors):
    print(f"Token: {token:12s} | Vector shape: {vec.shape}")
```

```python
from sentence_transformers import SentenceTransformer

text = "have the bards who preceded me left any theme unsung"

# Hugging Face 임베딩 모델 로드 (예: all-MiniLM-L6-v2, BAAI/bge-small-en-v1.5 등)
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

# 4. 텍스트를 고차원 벡터로 변환 (Vectorization)
vector = model.encode(text)
```
