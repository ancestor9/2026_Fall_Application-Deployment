```python
# @title Summary
import sys

def demo_memory_interaction():
    print("=== 1. 파이썬 변수와 메모리 주소(RAM) 확인 ===")

    # 1. 정수형 변수 생성 및 메모리 주소 확인
    a = 100
    b = 100
    print(f"a의 값: {a}, 메모리 주소(Hex): {hex(id(a))}")
    print(f"b의 값: {b}, 메모리 주소(Hex): {hex(id(b))}")
    print(f"a와 b는 같은 메모리를 가리키는가? -> {a is b}\n")
    # -> 파이썬은 소형 정수(일반적으로 -5~256)를 메모리에 미리 만들어 두고 재사용함 (메모리 절약)

    # 2. 값의 변경(연산)과 메모리 주소의 변화 (Immutable 객체)
    print("=== 2. CPU 연산에 따른 메모리 주소 변화 ===")
    original_address = id(a)
    a = a + 1  # CPU가 a의 값을 읽어와 1을 더한 후 '새로운 객체'를 생성함

    print(f"연산 후 a의 값: {a}")
    print(f"연산 전 주소: {hex(original_address)}")
    print(f"연산 후 주소: {hex(id(a))}")
    print("-> 정수 연산 시 기존 메모리 값이 바뀌는 것이 아니라, 새로운 메모리 주소에 결과가 생성됨!\n")

    # 3. 가변 객체(Mutable: List)에서의 메모리 참조 방식
    print("=== 3. List(가변 객체)에서의 메모리 참조 ===")
    list1 = [1, 2, 3]
    list2 = list1  # 주소 복사 (Shallow Copy)

    print(f"list1 주소: {hex(id(list1))}")
    print(f"list2 주소: {hex(id(list2))}")

    list1.append(4)
    print(f"list1 변경 후 list2의 내용: {list2}")
    print(f"list1 주소 유지 여부: {hex(id(list1))}")
    print("-> 리스트는 같은 메모리 주소를 공유하므로, 내부 값을 변경해도 메모리 주소가 유지됨!")

if __name__ == "__main__":
    demo_memory_interaction()
```

# **1. Arithmetic Operations(연산)**

Python implements seven basic binary arithmetic operators, two of which can double as unary operators.
They are summarized in the following table:

| Operator     | Name           | Description                                            |
|--------------|----------------|--------------------------------------------------------|
| ``a + b``    | Addition       | Sum of ``a`` and ``b``                                 |
| ``a - b``    | Subtraction    | Difference of ``a`` and ``b``                          |
| ``a * b``    | Multiplication | Product of ``a`` and ``b``                             |
| ``a / b``    | True division  | Quotient of ``a`` and ``b``                            |
| ``a // b``   | Floor division | Quotient of ``a`` and ``b``, removing fractional parts |
| ``a % b``    | Modulus        | Integer remainder after division of ``a`` by ``b``     |
| ``a ** b``   | Exponentiation | ``a`` raised to the power of ``b``                     |
| ``-a``       | Negation       | The negative of ``a``                                  |
| ``+a``       | Unary plus     | ``a`` unchanged (rarely used)                          |

# **2. Comparison Operations(비교연산)**
For this, Python implements standard comparison operators, which return Boolean values ``True`` and ``False``.
The comparison operations are listed in the following table:

| Operation | Description                   | Operation | Description                      |
| --------- | ----------------------------- | --------- | -------------------------------- |
| `a == b`  | `a` equal to `b`              | `a != b`  | `a` not equal to `b`             |
| `a < b`   | `a` less than `b`             | `a > b`   | `a` greater than `b`             |
| `a <= b`  | `a` less than or equal to `b` | `a >= b`  | `a` greater than or equal to `b` |


```python
import sys

def trace_memory_address_and_size():
    print("=== 1. 기본 자료형의 메모리 주소 및 용량 추적 ===")

    # 정수, 실수, 문자열, 불리언 예시
    sample_data = {
        "정수 0": 0,
        "정수 100": 100,
        "큰 정수 (2^60)": 2**60,
        "실수 3.14": 3.14,
        "문자열 'a'": "a",
        "문자열 'hello'": "hello",
        "불리언 True": True
    }

    for name, val in sample_data.items():
        addr = hex(id(val))             # 메모리 주소 (16진수)
        size = sys.getsizeof(val)       # 메모리 용량 (Byte)
        print(f"[{name:<12}] 주소: {addr} | 용량: {size:>3} Bytes")

    print("\n=== 2. 컨테이너 객체(리스트)의 메모리 특징 ===")

    empty_list = []
    print(f"빈 리스트 용량: {sys.getsizeof(empty_list)} Bytes")

    # 동적 배열 할당에 따른 메모리 크기 변화 추적
    dynamic_list = []
    print("\n[요소 추가에 따른 리스트 메모리 변화]")
    for i in range(5):
        dynamic_list.append(i)
        print(f"요소 개수: {len(dynamic_list)} | 주소: {hex(id(dynamic_list))} | 용량: {sys.getsizeof(dynamic_list)} Bytes")

    print("\n=== 3. 리스트의 '얕은 측정(Shallow Size)' 주의점 ===")

    # 64비트 환경에서 주소 포인터 하나는 보통 8 Bytes입니다.
    int_element = 10**20  # 큰 정수 객체
    list_with_large_int = [int_element]

    print(f"내부 정수 객체 자체 용량 : {sys.getsizeof(int_element)} Bytes")
    print(f"정수를 담은 리스트 전체 용량: {sys.getsizeof(list_with_large_int)} Bytes")
    print("-> 리스트 용량에는 내부 정수 객체의 실제 크기가 합산되지 않고, '포인터 용량'만 포함됩니다!")

if __name__ == "__main__":
    trace_memory_address_and_size()
```
