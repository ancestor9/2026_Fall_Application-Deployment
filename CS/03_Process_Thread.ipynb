```python
import os
from threading import Thread
from multiprocessing import Process

# 프로그램이 시작되면 메모리에 shared_val이 만들어지고 0을 가리킴
shared_val = 0
print(f'shared_val의 메모리 주소 : {id(shared_val)}')

# 프로세스(Process)의 ID 번호
print(f'shared_val의 프로세스 ID : {os.getpid()}')
```

```python
def add_val():
    '''
    프로세스(Process)의 ID 번호, 즉 PID(Process ID)를 반환하는 함수
    '''
    global shared_val
    shared_val += 10
    print(f"shared_vald의 프로세스 ID : [{os.getpid()}] 작업 후 shared_val: {shared_val}")
```

```python
# @title **1. 스레드: 동일 PID, 메모리 공유 -> shared_val = 10**
t = Thread(target=add_val)
t.start()
t.join()

# print(f"메인(Thread 실행 후): {shared_val}")  # 출력: 10
```

start() = 프로세스를 시작해라

join() = 프로세스가 끝날 때까지 기다려라

```python
# @title **2. 프로세스: 다른 PID, 메모리 독립 -> shared_val 복사본 변경**
shared_val = 0
print(f'shared_val의 메모리 주소 : {id(shared_val)}')
print(f'shared_val의 프로세스 ID : {os.getpid()}')
```

```python
p = Process(target=add_val)
p.start()
p.join()

print(f"메인(Process 실행 후): {shared_val}") # 출력: 0 (변화 없음)
```

```python
print(f'shared_val의 메모리 주소 : {id(shared_val)}')
print(f'shared_val의 프로세스 ID : {os.getpid()}')
```

### 서로 다른 Process

```python
def work():
    print(os.getpid())

def job():
    print(os.getpid())

if __name__ == "__main__":
    print("부모:", os.getpid())
    p = Process(target=work)
    p.start()
    p.join()

    p = Process(target=job)
    p.start()
    p.join()
```
