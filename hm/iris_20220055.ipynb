## <font color='orange'>  **1. Obtain Data**

```python
from sklearn.datasets import load_iris
import pandas as pd
```

```python
iris  = sklearn.datasets.load_iris()
df = pd.DataFrame(iris.data,
             columns=iris.feature_names
             )
df['target'] = iris.target
df.head()
```

```python
# target 이름을 문자열로
df['species'] = df['target'].map({0:'setosa', 1:'versicolor', 2:'virginica'})
df.head()
```

## Feature Engineering (Data Preprocessing)
-

```python
tf = df.copy()
tf
```

```python
# 스케일링할 특성(feature) 열 선택
feature_cols = ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']

# 직접 수식(Min-Max)을 사용하여 0과 1 사이로 변환
tf[feature_cols] = (tf[feature_cols] - tf[feature_cols].min()) / (tf[feature_cols].max() - tf[feature_cols].min())

# 결과 확인
tf.head()
```

```python
tf.describe()
```

```python
from sklearn.preprocessing import MinMaxScaler

# 스케일링할 특성(feature) 열 선택 (target과 species 제외)
feature_cols = ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']

# MinMaxScaler 객체 생성 및 적용
scaler = MinMaxScaler()
tf[feature_cols] = scaler.fit_transform(tf[feature_cols])

# 결과 확인
tf.head()
```
